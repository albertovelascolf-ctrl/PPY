Un **ordenador** es, en esencia, un **sistema** capaz de realizar cálculos y tomar decisiones lógicas a una velocidad increíblemente alta.

Dentro de estos sistemas distinguimos una parte física, **hardware**, formada por todos los componentes reales que conforman el dispositivo; y una parte lógica, **software**, compuesta por todos los datos que la máquina es capaz de **procesar y almacenar**. El ordenador procesará estos datos según el control realizado por una secuencia de instrucciones predefinidas a las que llamamos **programas**.

Al igual que podríamos utilizar distintos lenguajes (inglés, español, ...) para escribir un texto tradicional, existen muchos lenguajes de programación diferentes.

Un programa informático será entonces un "texto" que incluye las instrucciones que un ordenador va a necesitar ejecutar para poder resolver un problema

# 1. Algoritmos y diagrama de flujo

Un **algoritmo** es una serie de pasos a seguir para realizar una determinada tarea técnica.

Se puede hacer un simil con una receta de cocina. Una receta te dice como preparar un plato siguiendo una serie de pasos. Por ejemplo, para preparar un pastel los pasos a seguir serían: precalentar el horno, mezclar la harina con el azúcar y los huevos, poner la mezcla en un molde, ... y así sucesivamente.

Sin embargo, un algoritmo es un término técnico más específico que una simple receta. Para que algo sea considerado **algoritmo** debe de **cumplir** las siguientes **condiciones**:

1.   Debe **definir** de manera **clara** que debe hacerse en cada paso.
2.   Debe definir el **tipo de datos** que va a necesitar (entrada). Por ejemplo, se puede requerir de dos números y que ambos sean mayores que cero. O puede requerir de una palabra, o una lista de tres números.
3. Debe **generar un resultado** (salida) concreto y definido. Por ejemplo, puede generar el número que sea mayor de dos entradas, la versión en mayúscula de una palabra o una lista ordenada de números.
4. Debe **terminar** y producir un resultado en un **tiempo finito**.
5. Un algoritmo ha de ser capaz de generar el **resultado correcto en cualquier situación**.
6. Si un algoritmo impone una condición a las entradas, esa condición se debe satisfacer.

La representación de un algoritmo a través de texto puede ser tediosa y difícil. Por eso se utilizan los diagramas de flujo. **Un diagrama de flujo** será entonces, una representación gráfica mendiante símbolos de un algoritmo.

# 2. Estructura de un programa
**Programar** es transformar esos algoritmos que tú entiendes como persona, en instrucciones que puede entender el ordenador. Por lo tanto, un **programa** es un conjunto de instrucciones que siguen un orden lógico y que están escritas en un lenguaje de programación.

Igual que un algoritmo, un programa sigue esta estructura:

- **Entrada:** el programa recibe datos (por ejemplo, lo que escribe una persona).

- **Proceso:** el programa hace algo con esos datos (una operación, una comparación…).

- **Salida:** el programa muestra un resultado (por ejemplo, en pantalla).

# 4. Mostrar información

Vamos a comenzar por algo básico pero fundamental: mostrar información por pantalla. Esto es lo que normalmente ocurre al final de un algoritmo: comunicar el resultado. Para ello, usaremos una función muy importante en Python:

```python
print("Hola Mundo")
```

La función `print()` sirve para mostrar por pantalla cualquier texto, número o resultado que quieras que aparezca. Todo lo que pongas dentro de `print()` se mostrará en la pantalla cuando ejecutes la celda. Cada `print()` muestra el texto en una línea nueva.

**¿Qué se puede mostrar por pantalla?**
- **Texto:** El texto siempre va entre comillas: `" "` o `' '`. A eso se le llama una **cadena de caracteres** o **cadena de texto**.



```python
print("Esto es un mensaje")
```

- **Números:** Se pueden mostrar números enteros y decimales. La coma decimal se representa con un punto (por ejemplo, `3.14`).

```python
print(42)
print(3.14)
```

- **Resultados:** Python calcula el resultado de operaciones y funciones antes de mostrarlo. (La función `len()` nos devuelve la cantidad de caracteres en una cadena de texto).

```python
print(1 + 1)
print(8 - 3)
print(len("Hola mundo"))
```

- **Varias cadenas a la vez:** Para unir dos cadenas de texto usamos la operación **concatenar** `+`.

```python
print("Hola " + "mundo")
```

- **Varios valores a la vez:** En Python, puedes separar los elementos dentro del `print()` con comas para mostrar varios datos.

```python
print("El resultado es", 5 + 3)
```

# 5. Variables y tipos de datos

Una variable es un espacio de memoria reservado en nuestro programa para almacenar un valor. Este valor puede ser un número, una cadena de texto, un objeto, entre otros. En Python, las variables se crean al darles un nombre y asignarles un valor. Por ejemplo, si queremos crear una variable llamada “edad” que almacene un valor de 25, podemos escribir:

```python
edad = 25
```

Ahora, podemos utilizar esta variable en nuestro programa. Por ejemplo, podríamos imprimir el valor de la variable utilizando la función `print()`. En este caso nos aparecerá por pantalla el valor 25.

```python
print(edad)
```

Más adelante, si queremos cambiar el valor almacenado en nuestra variable “edad”, podemos hacerlo simplemente asignándole un nuevo valor. Si hiciesemos un `print()`de "edad", nos aparecerá el valor actualizado

```python
edad = 30
print(edad)
```

Es importante tener en cuenta que en Python, las variables son dinámicas, lo que significa que no necesitamos especificar de antemano qué tipo de valor almacenarán. Por ejemplo, podríamos crear una variable llamada “nombre” que almacena una cadena de texto:

```python
nombre ='Juan'
print(nombre)
```

Y luego, cambiarla a almacenar un número:

```python
nombre = 123
print(nombre)
```

Aunque esta flexibilidad puede ser conveniente, también puede hacer que el código sea más difícil de leer y mantener. Para ayudar a mitigar esto, es útil seguir buenas prácticas de programación, como nombrar las variables de manera **descriptiva**, es decir que el nombre de la variable describa su proposito y finalidad.

A continuación vamos a poner ejemplos de nombres de variables que **NO SON DESCRIPTIVOS** y que no deberías utilizar:

```python
numero = "palabra"
numerito = 12
asdasdfa = "Error en el login"
p = 23
numeroqueutilizoparacomprobarsihayalgunerror = 1
```

Ejemplos de variables que **SÍ** son descriptivas:

```python
num1 = 12
mensaje_error_login = "Error en el login"
x = 1
y = 2
```

**NOTA IMPORTANTE:** las variables, por convenio, siempre van en minúscula y si tienes que poner algún espacio pondrás el carácter "_"para representarlo (como hicimos con la variable mensaje_de_error). Además siempre tiene que empezar con letra.

Hay que decir que en Python hay ciertas palabras que están **reservadas** y no se pueden utilizar como nombres de variables. Palabras como `True, False, def, str, int`, y muchas más no se pueden utilizar para declarar una variable ya que tienen otro uso en Python.

# 7. Constantes

Una **constante** es un valor que **no debería cambiar** durante la ejecución del programa.
Python no tiene constantes verdaderas, pero por convenio (es decir, por costumbre),
se usan **nombres en mayúsculas** para indicar que un valor no se va a modificar.

```python
PI = 3.14159
print("Perímetro de una circunferencia de radio 2:", 2 * PI * 2)
```

**¿Por qué usar constantes?**

- Para que el código sea más claro y fácil de mantener.

- Para evitar errores si ese valor se usa en muchos sitios.

- Para dar más sentido al programa (es más claro escribir PI que 3.14159 en varios sitios).

# 10. Entrada de datos
Al igual que la función `print()` sirve para mostrar información, disponemos de la función `input()` para introducir datos en el programa para que puedan ser procesados. Estos datos se
almacenan en una variable y se utilizan cuando el programa los necesita en función de las tareas que deba llevar a cabo.


```python
variable = input("Mensaje para solicitar el dato: ")
print(variable)
n = input("Escribe un número: ")
print(n)
```

La función `input()` devuelve siempre una cadena de texto, por lo que, si queremos utilizar el valor introducido como número será necesario realizar una conversión de datos.

**Sin** conversión de datos:

```python
n = input("Escribe un número: ")
print(n + 5)
```

**Con** conversión de datos:

```python
n = input("Escribe un número: ")
print(int(n) + 5)

n = int(input("Escribe un número: "))
print(n + 5)
```
