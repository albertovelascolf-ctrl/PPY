```python
jugador_vida = 100
jugador_ataque = 20

enemigo_vida = 80
enemigo_ataque = 15

print("¡Empieza el combate!")
while jugador_vida > 0 and enemigo_vida > 0:
    input("Pulsa para atacar")
    enemigo_vida -= jugador_ataque
    print(f"Atacas al enemigo. Vida enemigo {enemigo_vida}")
    if enemigo_vida <= 0:
        break
    jugador_vida -= enemigo_ataque
    print(f"El enemigo ataca. Tu vida {jugador_vida}")

if jugador_vida > 0:
    print("¡Has ganado!")
else:
    print("Has perdido...")
```

```python
class Personaje:
    def __init__(self, nombre, vida, ataque):
        self.nombre = nombre
        self.vida_maxima = vida
        self.vida_actual = self.vida_maxima
        self.ataque = ataque

    def __str__(self):
        return f"{self.nombre}: Vida {self.vida_actual} - Ataque {self.ataque}"

    def atacar(self, otro):
        otro.vida_actual -= self.ataque
        print(f"{self.nombre} ataca a {otro.nombre}")
        print(f"A {otro.nombre} le quedan {otro.vida_actual} puntos de vida")

    def curar(self):
        print(f"{self.nombre} se cura")
        self.vida_actual += 25
        if self.vida_actual > self.vida_maxima:
            perdida = self.vida_actual - self.vida_maxima
            self.vida_actual = self.vida_maxima
            print(f"{self.nombre} ha desaprovechado {perdida} puntos de vida")
        print(f"A {self.nombre} le quedan {self.vida_actual} puntos de vida")

    def esta_vivo(self):
        return self.vida_actual > 0

jugador = Personaje("Héroe", 100, 20)
enemigos = [
    Personaje("Goblin", 50, 10),
    Personaje("Orco", 80, 15),
    Personaje("Troll", 100, 20)
]

enemigo = enemigos.pop(0)
print(f"¡Nuevo enemigo! Ha aparecido {enemigo}")

while jugador.esta_vivo() and (enemigos or enemigo.esta_vivo()):
    accion = input("Elige tu acción: \n1 - Atacar \n2 - Curarse")
    if accion == "1":
        jugador.atacar(enemigo)
    elif accion == "2":
        jugador.curar()
    else:
        print("Acción no válida")
        continue
    if not enemigo.esta_vivo():
        jugador.curar()
        if enemigos:
            print(f"{enemigo.nombre} ha perdido toda la vida")
            enemigo = enemigos.pop(0)
            print(f"¡Nuevo enemigo! Ha aparecido {enemigo}")
            continue
        else:
            break
    enemigo.atacar(jugador)

if jugador.esta_vivo():
    print("¡Has ganado!")
else:
    print("Has perdido...")
```
