En Python, las **estructuras de datos** son la forma en que organizamos y **almacenamos** datos de manera **eficiente**, permitiendo manipular y procesar información de forma organizada. Las estructuras de datos comunes en Python incluyen cadenas de caracteres, listas, tuplas y diccionarios.

## Cadenas de caracteres

Un string es una secuencia de caracteres **inmutable** (es decir, que no se puede cambiar una vez declarado) entre comillas. Puedes usar comillas simples ' o dobles " indistintamente:


````python
cadena1= "hola"
cadena2= 'hola'
````

Una vez tenemos un string declarado, podemos acceder a los caracteres individuales de dicha cadena. Para ello tenemos que utilizar los índices.


Un índice es una posición en una cadena. El primer elemento de un string se encuentra en la posición 0 de la cadena y el último elemento se encuentra en la posición n-1 (siendo n la longitud de la cadena).


| Caracter | P | y | t | h | o | n |
| -------- | - | - | - | - | - | - |
| Índice   | 0 | 1 | 2 | 3 | 4 | 5 |

Ahora veremos como acceder a cada uno de los índices en Python:


````python
cadena= "Python"
print(cadena[0])
print(cadena[1])
print(cadena[2])
print(cadena[3])
print(cadena[4])
print(cadena[5])
````

Python tiene la peculiaridad de poder usar índices negativos para acceder a los caracteres.

| Carácter | P | y | t | h | o | n |
| -------- | - | - | - | - | - | - |
| Índice   | 0 | 1 | 2 | 3 | 4 | 5 |
| Índice Negativo | -6 | -5 | -4 | -3 | -2 | -1 |

De esta forma es equivalente acceder al string en la posición 0 o -6, como también lo es la 5 y la -1.

````python
print(cadena[-1])
print(cadena[-2])
print(cadena[-3])
print(cadena[-4])
print(cadena[-5])
print(cadena[-6])
````

**¡CUIDADO!** Cuando creamos una cadena de caracteres, se reserva en memoria el espacio justo para esa variable. Esto quiere decir que si la variable tiene una longitud de 5 caracteres e intentamos acceder a un índice superior al reservado, va a dar error.

````python
print(cadena[6]) # El error que nos va a salir es IndexError: string index out of range
````

### Slicing

En Python puedes usar rangos de índices para trabajar con subcadenas dentro de un string. Esta técnica se llama **slicing** (segmentación) y tiene la siguiente forma:


```
cadena[inicio:fin]
```
Siendo:

- `inicio`: posición donde empieza (se incluye).
- `fin`: posición donde termina (no se incluye).

Veamos el siguiente ejemplo:

````python
texto = "Miniyoda"

print(texto[0:4])  # Mini (índices 0, 1, 2 y 3)
print(texto[4:8])  # yoda (índices 4, 5, 6 y 7)
print(texto[:])    # Miniyoda (copia completa del string)
print(texto[:5])   # Miniy (desde el principio hasta el índice 4)
print(texto[2:])   # niyoda (desde el índice 2 hasta el final)
print(texto[-3:])  # oda (últimos tres caracteres)
print(texto[:-4])  # Mini (todos excepto los cuatro últimos)
````

También se puede incluir como tercer parámetro opcional un paso o salto

````python
print(texto[::2])  # Mnyd (salta de 2 en 2)
print(texto[::-1]) # adoyiniM (cadena al revés)
````

### Funciones con cadenas de caracteres

Las cadenas de caracteres en Python ofrecen herramientas muy útiles para gestionar y transformar su contenido. Estas herramientas incluyen funciones y métodos que permiten realizar operaciones como:

- Obtener información.
- Cambiar contenido.
- Dividir y unir cadenas.
- Buscar.
- Eliminar espacios y caracteres.
- Comprobaciones lógicas.
- Formatear cadenas.

Ahora iremos viendo de forma detallada cada una de las funciones básicas para operar y manejar dichas cadenas de caracteres:

- **`len(string)`**: Devuelve la longitud de la cadena (número de caracteres).



````python
cadena_ejemplo = "Python"
print(len(cadena_ejemplo))  # 6
````

- **`string1 in string2`**: Verifica si una subcadena está dentro de otra.

````python
print("Py" in "Python")  # True
print("Tutti" in "Fruti")  # False
````

- **`.lower()`** y **`.upper()`**: estas dos funciones sirven para poner una cadena de caracteres en minúscula o mayúscula.

````python
print(cadena_ejemplo.lower())  # python
print(cadena_ejemplo.upper())  # PYTHON

# También se pueden usar strings directamente

print("Hola".lower())  # hola
print("Hola".upper())  # HOLA
````

- **`.replace(subcadena1, subcadena2)`**: reemplaza una subcadena por otra.

````python
saludo = "Hola Mundo"
print(saludo.replace("Mundo", "Python"))  # Hola Python

#También se puede hacer con strings directamente
print("Hola mundo".replace("mundo", "Python"))  # Hola Python
````

- **`.find(subcadena)`**: Devuelve el índice de la primera aparición (o -1 si no está).

````python
print("Python".find("o"))  # 4
print("Python".find("th")) # 2
print("Python".find("z"))  # -1
````

- **`.strip(), .lstrip(), .rstrip()`**: Eliminan espacios al principio y/o final.

````python
texto = "  hola  "
print(texto.strip())   # "hola"
print(texto.lstrip())  # "hola  "
print(texto.rstrip())  # "  hola"
````

- **Comprobaciones lógicas:** Estas devuelven True o False según el contenido del string.

````python
cadena = "123"

print(cadena.isdigit())   # True (solo números)
print(cadena.isalpha())   # False (no solo letras)
print(cadena.isalnum())   # True (letras y números)
print("hola".islower())   # True
print("HOLA".isupper())   # True
print(" ".isspace())      # True
````

- **Formatear strings**: a la hora de usar variables dentro de un print, podemos usarlas como hemos visto hasta ahora o bien formateandolas usando lo que se conoce como f-strings.

````python
nombre = "Ana"
edad = 30
print(f"{nombre} tiene {edad} años")  # Ana tiene 30 años
````

A modo de resumen, estas son las funciones más importantes:

| Función | Acción |
| -------- | - |
| `len(string)`   | Número de caracteres |
| `in`   | Comprueba si una subcadena está |
| `lower()`   | String en minúsculas |
| `upper()`   | String en mayúsculas |
| `replace(string1, string2)`   | Remplaza una cadena con otra |
| `find(string)`   | Devuelve el índice donde empieza la subcadena |
| `strip()`   | Elimina espacios dentro de un string |
| `lstrip()`   | Elimina espacios al principio del string |
| `rstrip()`   | Elimina espacios al final del string |

## Listas

Una **lista** es una estructura de datos que permite almacenar varios valores en una sola variable. Se utiliza para guardar colecciones de elementos, que pueden ser del mismo tipo o de tipos distintos.

Para crear una lista, se usan corchetes `[]` y se escriben los elementos separados por comas.

````python
numeros = [10, 20, 30, 40, 50]
nombres = ["Ana", "Luis", "Carlos"]
mezcla = [1, "hola", 3.14]
vacia = []
````

Al igual que pasa con las cadenas de caracteres, cada elemento de la lista tiene un **índice**, empezando desde 0, y también se pueden usar **índices negativos** para acceder desde el final hacia el principio de la lista.

| Lista | [ 4, | 2, | "a", | True ] |
| -------- | - | - | - | - |
| Índice   | 0 | 1 | 2 | 3 |
| Índice negativo   | -4 | -3 | -2 | -1 |

````python
frutas = ["manzana", "naranja", "plátano", "uva"]

print(frutas[0])
print(frutas[2])
print(frutas[-1])
print(frutas[-3])
````

A diferencia de las cadenas de caracteres, que son inmutables, las listas sí que se pueden modificar. Para modificar un elemento de una lista debemos indicar el índice del elemento y su nuevo valor.

En el siguiente ejemplo, modificaremos el valor del elemento en el índice 1 (`"naranja"`) por el nuevo valor, `"mandarina"`.

````python
frutas[1] = "mandarina"
print(frutas)
````

Si queremos trabajar con cada uno de los elementos de una lista de forma independiente podemos apoyarnos en el bucle `for`.

La sintaxis general es:

```
for elemento in lista:
    # hacer algo con elemento
```

Aquí, en cada vuelta del bucle, la variable elemento toma el valor de un elemento diferente de la lista, desde el primero hasta el último.

Dentro del bucle, puedes hacer cualquier operación con el elemento actual, por ejemplo:

- Modificarlo (si es mutable o asignarlo a otra variable),
- Realizar cálculos,
- Concatenar cadenas,
- Usarlo en condiciones, etc.

````python
numeros = [5, 7, 6, 8, 3, 4, 1]

for numero in numeros:
  if numero % 2 == 0:
    print(numero, "es par")
  else:
    print(numero, "es impar")
````

También podemos utilizar el **slicing** de la misma forma que con las cadenas de caracteres.

````python
numeros = [10, 20, 30, 40, 50, 60, 70, 80]

print(numeros[0:4])  # [10, 20, 30, 40] (índices 0, 1, 2, 3)
print(numeros[4:8])  # [50, 60, 70, 80] (índices 4, 5, 6, 7)
print(numeros[:])    # la lista completa
print(numeros[:6])   # desde el primer elemento hasta el índice 5
print(numeros[::-1]) # la lista al revés
````

**¡IMPORTANTE!** El slicing no modifica la lista original. Si quieres guardar el resultado, debes asignarlo a una nueva variable.

### Funciones con listas
Las listas en Python no solo permiten almacenar múltiples valores en una sola variable, sino que también ofrecen herramientas muy útiles para gestionar y transformar su contenido. Estas herramientas incluyen funciones y métodos que permiten realizar operaciones como:

- Contar elementos,
- Añadir nuevos valores,
- Eliminar o buscar datos,
- Ordenar o invertir la lista,
- Calcular totales, máximos, mínimos, etc.

Estas operaciones son esenciales para resolver problemas cotidianos en programación, ya que permiten trabajar con los datos de forma flexible y eficiente.

- **`len(lista)`:** Devuelve el número de elementos que contiene la lista.

````python
numeros = [40, 60, 10, 30]
print(len(numeros))
````

- **`elemento in lista`:** Devuelve si un elemento está en una lista.

````python
frutas = ["naranja", "plátano", "manzana"]
print("naranja" in frutas)
print("pera" in frutas)
````

- **`sum(lista)`:** Devuelve la suma de todos los elementos numéricos de la lista.

````python
numeros = [40, 60, 10, 30]
print(sum(numeros))
````

- **`max(lista)`y `min(lista)`:** devuelven el valor más grande y más pequeño de la lista.

````python
numeros = [40, 60, 10, 30]
print(max(numeros))
print(min(numeros))

letras = ["d", "a", "z", "c"]
print(max(letras))
print(min(letras))
````

- **`sorted(lista)`:** devuelve una nueva lista ordenada, sin modificar la original. Si queremos ordenarla en orden descendente, añadimos `reverse=True`.

````python
valores = [4, 2, 7, 1]
ordenados = sorted(valores)
print(ordenados)      # Resultado: [1, 2, 4, 7]
print(valores)        # La original sigue igual: [4, 2, 7, 1]

ordenados_desc = sorted(valores, reverse=True)
print(ordenados_desc)
````

- **`cadena.split(separador)`**: Divide el string en una lista usando un separador.

````python
frase = "uno,dos,tres"
print(frase.split(","))  # ['uno', 'dos', 'tres']
````

-  **`cadena.join(lista)`**: Une los elementos de una lista con un separador.

````python
numeracion = ["uno", "dos", "tres"]
print(",".join(numeracion))  # uno,dos,tres
````

- **`lista.append(elemento)`:** Añade un elemento al final de la lista.

````python
frutas = ["manzana", "plátano"]
frutas.append("naranja")
print(frutas)
````

- **`lista.insert(posición, elemento)`:** Inserta un elemento en una posición concreta. Los demás elementos se desplazan.

````python
numeros = [1, 2, 4]
numeros.insert(2, 3)  # Inserta el 3 en la posición 2
print(numeros)
````

- **`lista.remove(elemento)`:** Elimina la primera aparición de un elemento de la lista. Si no existe, da error.

````python
colores = ["rojo", "verde", "azul", "verde"]
colores.remove("verde")
print(colores)
````

- **`lista.pop(índice)`:** Elimina un elemento y lo devuelve. Si no se indica índice, elimina el último elemento.

````python
letras = ["a", "b", "c"]
ultimo = letras.pop()
print(ultimo)
print(letras)

primero = letras.pop(0)
print(primero)
print(letras)
````

- **`lista.clear()`:** Elimina todos los elementos de la lista.

````python
nombres = ["Ana", "Luis", "Marta"]
nombres.clear()
print(nombres)
````

- **`lista.count(elemento)`:** Devuelve cuántas veces aparece un elemento en la lista.

````python
notas = [7, 5, 7, 6, 7]
print(notas.count(7))
````

- **`lista.index(elemento)`:** Devuelve el índice de la primera aparición de elemento en la lista. Si no existe, lanza un error.

````python
dias = ["lunes", "martes", "miércoles"]
print(dias.index("martes"))
````

- **`lista.sort()`:** Ordena los elementos de la lista en orden. Modifica la lista. Se puede usar `reverse=True` para orden descendente.

````python
numeros = [3, 1, 4, 2]
numeros.sort()
print(numeros)

numeros.sort(reverse=True)
print(numeros)
````

A modo de resumen, estas son las funciones más importantes:

| Función | Acción |
| -------- | - |
| `len(lista)`   | Número de elementos |
| `elemento in lista`   | Indica si existe un elemento |
| `sum(lista)`   | Suma de elementos numéricos |
| `max(lista)`   | Valor máximo |
| `min(lista)`   | Valor mínimo |
| `sorted(lista)`   | Nueva lista ordenada |
| `cadena.split(separador)`   | Nueva lista a partir de una cadena |
| `cadena.join(lista)`   | Nueva cadena a partir de una lista |
| `lista.append(elemento)`   | Añade al final |
| `lista.insert(posición, elemento)`   | Inserta un una posición || `lista.remove(elemento)`   | Elimina la primera aparición || `lista.remove(índice)`   | Elimina y devuelve un elemento || `lista.clear()`   | Vacía la lista |
| `lista.index(elemento)`   | Índice de la primera aparición de un elemento |
| `lista.count(elemento)`   | Cuenta apariciones de  elemento |
| `lista.sort()`   | Ordena la lista |

## Tuplas

Una tupla es un tipo de estructura de datos en Python que sirve para almacenar varios elementos en una sola variable, igual que una lista. La diferencia principal es que las tuplas son **inmutables**, es decir, no se pueden modificar una vez creadas. Además son ordenadas y pueden contener distintos tipos de datos.

Para definir una tupla se hace de la siguiente forma:

````python
mi_tupla = (1, 2, 3, 4, 5)
tu_tupla = ("a", 30, "c", 50)
print(mi_tupla)
print(tu_tupla)
````

Si quieres definir una tupla con un solo elemento se tiene que hacer de la siguiente forma:

````python
otra_tupla = (1,)
print(otra_tupla)
````

Como las listas, se puede acceder a los elementos de una tupla por su índice:

````python
print(mi_tupla[0])
print(tu_tupla[2])
````

Como las tuplas son inmutables, no podemos modificar el valor de un elemento. Si quisiesemos hacerlo, deberíamos convertir la tupla en una lista, modificar el valor y luego desconvertirla.

````python
mi_tupla = (1, 2, 3, 4, 5)
mi_lista = list(mi_tupla)
mi_lista[0] = 10
mi_tupla = tuple(mi_lista)
print(mi_tupla)
````

## Diccionarios
Los **diccionarios** son una estructura de datos que permite almacenar pares de información, en forma de **`clave: valor`**. Se utilizan cuando necesitas acceder a un dato asociado a una clave específica, en lugar de usar índices como en las listas.

Un diccionario se define entre llaves `{}` y cada elemento está formado por una clave y un valor separados por dos puntos. En este ejemplo, "nombre", "edad" y "ciudad" son **claves**, mientras que "Lucía", 25 y "Valencia" son sus respectivos **valores**.



````python
persona = {
    "nombre": "Lucía",
    "edad": 25,
    "ciudad": "Valencia"
}

print(persona)
````

Podemos acceder al valor asociado a una clave usando corchetes `[]`:

````python
print(persona["nombre"])
print(persona["edad"])
````

Si intentamos acceder a una clave que no existe, Python mostrará un error (`KeyError`). Para evitarlo, podemos usar el método `.get()` que devuelve `None` si no encuentra la clave:

````python
print(persona.get("altura"))
````

Podemos añadir una nueva clave o modificar el valor de una clave existente simplemente asignando un nuevo valor:

````python
persona["edad"] = 26                # Modifica la edad
persona["altura"] = 1.70            # Añade una nueva clave

print(persona)
````

También podemos recorrer el diccionario con un bucle `for`, pero solo nos mostrará las claves del diccionario.

````python
for clave in persona:
  print(clave)
````

### Funciones con diccionarios
- **`diccionario.get(clave)`:** Devuelve el valor asociado a una clave.


````python
print(persona.get("altura"))
````

-**`clave in diccionario`:** Devuelve su una clave existe en un diccionario.

````python
print("profesión" in persona)
print("ciudad" in persona)
````

- **`del diccionario[clave]`:** Borra una clave y su valor.

````python
del persona["altura"]
````

- **`diccionario.pop(clave)`:** Devuelve el valor asociado a una clave y borra dicha clave y su valor.

````python
edad = persona.pop("edad")
print(edad)
````

- **`diccionario.keys()`:** Devuelve las claves de un diccionario. Se usa habitualmente en los bucles para recorrer el diccionario.

````python
for clave in persona.keys():
  print(clave)
````

- **`diccionario.values()`:** Devuelve los valores de un diccionario.

````python
for valor in persona.values():
  print(valor)
````

- **`diccionario.items()`:** Devuelve la tupla `clave, valor`.

````python
for clave, valor in persona.items():
  print(clave, "->", valor)
````

- **`diccionario.clear()`:** Vacía el diccionario.

````python
persona.clear()
print(persona)
````

A modo de resumen, estas son las funciones más importantes:

| Función | Acción |
| -------- | - |
| `diccionario.get(clave)`   | Devuelve el valor asociado a la clave |
| `clave in diccionario`   | Devuelve si una clave existe |
| `del diccionario[clave]`   | Borra una clave y su valor |
| `diccionario.pop(clave)`   | Devuelve el valor asociado a la clave y borra la clave y su valor |
| `diccionario.keys()`   | Devuelve las claves |
| `diccionario.values()`   | Devuelve los valores |
| `diccionario.items()`   | Devuelve la tupla `clave, valor` |
| `diccionario.clear()`   | Vacía el diccionario |
