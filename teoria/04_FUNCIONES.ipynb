## Funciones
Una **función** es un bloque de código que realiza una tarea específica. Puede recibir datos (llamados **parámetros**), procesarlos, y opcionalmente devolver un resultado. Las funciones permiten reutilizar código y evitar repetir instrucciones.

En el siguiente ejemplo, `print()` es una función ya existente en Python, que muestra información por pantalla. Le pasamos un parámetro: el texto `"Hola mundo"`.

````python
print("Hola mundo")
````

Usar funciones en un programa tiene muchas ventajas:

- **Organización del código**: divide el programa en partes pequeñas y claras.
- **Reutilización:** una vez creada, una función se puede usar muchas veces.
- **Evita repetir código:** si una tarea se repite, solo hace falta escribirla una vez dentro de una función.
- **Facilita el mantenimiento y pruebas:** es más fácil localizar errores o hacer cambios.
- **Mejora la legibilidad:** un código dividido en funciones es más fácil de entender.

## Librerías
Una **librería** (o **módulo**) en Python es un conjunto de funciones, clases y constantes ya programadas que puedes importar para no tener que escribir todo desde cero. Existen:

- **Librerías estándar:** que ya vienen incluidas con Python.
- **Librerías externas:** que se pueden instalar.

Para importar una librería tenemos tres formas:
- Importar toda la librería:


````python
import math
print(math.sqrt(16))
````

- Importar con alias:

````python
import math as m
print(m.sqrt(16))
````

- Importar una parte concreta:

````python
from math import sqrt
print(sqrt(16))
````

### Librería `random`
La librería `random` se utiliza para generar números aleatorios o seleccionar elementos al azar. Es muy útil para:

- Simular situaciones (juegos, sorteos, experimentos).
- Crear datos de prueba.
- Introducir variabilidad en un programa.

>| Función                      | ¿Qué hace?                                                                | Mini ejemplo                               |
| ---------------------------- | ------------------------------------------------------------------------- | ------------------------------------------ |
| `random.random()`            | Devuelve un número decimal aleatorio entre 0 y 1 (sin incluir 1).         | `random.random() → 0.738...`               |
| `random.randint(a, b)`       | Devuelve un entero entre `a` y `b` (ambos incluidos).                     | `random.randint(1, 6) → 4`                 |
| `random.uniform(a, b)`       | Devuelve un decimal entre `a` y `b`.                                      | `random.uniform(1, 5) → 3.42`              |
| `random.choice(lista)`       | Elige un elemento aleatorio de una lista o cadena.                        | `random.choice(["rojo", "azul"]) → "rojo"` |
| `random.choices(lista, k=n)` | Elige `n` elementos al azar (con repetición).                             | `random.choices(["a", "b", "c"], k=2)`     |
| `random.shuffle(lista)`      | Mezcla los elementos de una lista **en el mismo objeto** (modifica).      | `random.shuffle(cartas)`                   |
| `random.sample(lista, k=n)`  | Elige `n` elementos **sin repetición**.                                   | `random.sample([1, 2, 3, 4], k=2)`         |
| `random.seed(x)`             | Establece una semilla para repetir resultados aleatorios (útil en tests). | `random.seed(42)`                          |


````python
import random

opciones = ["piedra", "papel", "tijera"]
eleccion = random.choice(opciones)
print("El ordenador ha elegido:", eleccion)
````

### Librería `datetime`

La librería `datetime` permite trabajar con fechas, horas, intervalos de tiempo y formatos temporales. Es útil en programas que necesiten:

- Saber la fecha u hora actual.
- Calcular diferencias entre fechas.
- Mostrar fechas en formatos personalizados.
- Controlar vencimientos, cumpleaños, agendas, etc.

> | Elemento                             | ¿Qué hace?                                                         | Mini ejemplo                                  |
| ------------------------------------ | ------------------------------------------------------------------ | --------------------------------------------- |
| `date.today()`                       | Devuelve la fecha actual.                                          | `date.today() → 2025-06-16`                   |
| `datetime.now()`                     | Devuelve la fecha y hora actuales.                                 | `datetime.now() → 2025-06-16 14:25:10.3456`   |
| `datetime(year, m, d, h, m)`         | Crea una fecha y hora personalizada.                               | `datetime(2023, 12, 25, 10, 30)`              |
| `timedelta(days=x)`                  | Representa un período de tiempo (p.ej. días, horas, minutos...).   | `fecha + timedelta(days=5)`                   |
| `fecha.year` / `.month` / `.day`     | Extrae año, mes o día de una fecha.                                | `hoy.year → 2025`                             |
| `fecha.strftime(formato)`            | Devuelve la fecha en un formato de texto personalizado.            | `hoy.strftime("%d/%m/%Y") → '16/06/2025'`     |
| `datetime.strptime(cadena, formato)` | Convierte un texto en una fecha según un formato.                  | `datetime.strptime("10-01-2023", "%d-%m-%Y")` |
| `fecha1 - fecha2`                    | Devuelve un objeto `timedelta` con la diferencia entre dos fechas. | `(hoy - otra_fecha).days`                     |


````python
from datetime import date, timedelta

hoy = date.today()
plazo = timedelta(days=7)
entrega = hoy + plazo

print("Hoy es:", hoy)
print("La entrega es dentro de 7 días:", entrega)
````

Y ¿qué pasa si **no uso** `return` en mis funciones? La función **no devuelve ningún valor**. Solo hace algo (como imprimir por pantalla), pero no puede entregar un resultado para guardarlo o usarlo en otras operaciones.

A la hora de declarar una función hay una serie de reglas que debemos de respetar:

- Usa nombres descriptivos.
- Empiezar con letra el nombre de la función (no con número) y escrito en minúsculas.
- No usar palabras reservadas como `print`, `if`, `while`, etc.
- Separa palabras con _ si es necesario (calcular_total, mostrar_mensaje).

## Recursividad

La recursividad es una técnica de programación en la que una función se llama a sí misma para resolver un problema.

Es como una caja que contiene otra caja igual dentro: cada llamada a la función es una "caja" más hasta que se llega a una condición de parada.

Las partes básicas de una función recursiva son:

- Caso base (condición de parada): cuándo detenerse.
- Llamada recursiva: cuando la función se llama a sí misma con un nuevo valor.

Ahora vamos a programar una función recursiva. El caso más típico es la función factorial. El factorial de un número se calcula de la siguiente forma:

```
Factorial del numero 8! = 8x7x6x5x4x3x2x1
Factorial del número N! = Nx(N-1)x...x2x1
```

````python
def factorial(n):
    if n == 0:
        return 1  # Caso base
    else:
        return n * factorial(n - 1)  # Llamada recursiva

print(factorial(4))
````

Lo que está pasando cuando se invoca a la función `factorial()`es

```
factorial(4)
4 * factorial(3)
4 * (3 * factorial(2))
4 * (3 * (2 * factorial(1)))
4 * (3 * (2 * (1 * factorial(0))))
4 * 3 * 2 * 1 * 1 = 24
```

**¡IMPORTANTE!** Si olvidas el caso base, tendrás una recursión infinita que dará error `(RecursionError)`
