En un lenguaje de programación, las estructuras de control permiten modificar el flujo de la ejecución de un conjunto de instrucciones. Se pueden distinguir tres tipos básicos de control de flujo, que son:

- **Control secuencial:** las instrucciones se ejecutan de manera consecutiva desde el inicio hasta el fin del programa.
- **Control de selección:** se tiene una condición que puede ser falsa o verdadera, dependiendo de esto se ejecutará uno u otro bloque de instrucciones.
- **Control de repetición:** un bloque de instrucciones se ejecuta de manera repetitiva mientras una condición sea verdadera, en caso contrario el flujo de ejecución se pasará a otro conjunto de instrucciones.

## Condicional `if-elif-else`

El condicional `if-elif-else` es una estructura de control de **selección** que sirve para tomar decisiones, basándose en la evaluación de condiciones y/o comparaciones, en el programa. La sintaxis más general para `if-elif-else` es:

```
if cond1:
    # hacer algo
elif cond2:
    # hacer otra cosa
...
elif condn:
    # hacer algo más
else:
    # hacer algo por defaul
```

Donde `cond1, cond2, ... condn` son valores lógicos que resultan de una comparación. Esta estructura se evalúa secuencialmente hasta encontrar una condición que se cumpla, si ninguna lo hace, entonces se ejecuta la instrucción colocada en el caso por default `else`.

Hay que tener especial cuidado con las **indentaciones** de los bloques de código, en Python las indentaciones no son opcionales, tienen un **significado sintáctico**. La mayoría de los editores de código de Python indentarán de manera automática la siguiente línea, cada vez que coloques los dos puntos al final de una línea.

En el siguiente fragmento de código se muestra un programa que verifica si el valor de `a` es mayor, menor o igual al valor de `b`.

````python
a = 10
b = 30

if a > b:
    print("a es mayor que b")
elif a < b:
    print("a es menor que b")
else:
    print("a es igual a b")
````

Podemos observar que en este caso dado que el valor de **a es menor b**, entonces se ejecuta la instrucción contenida en el bloque del elif.

Es muy común que en algunos casos, por cuestiones inherentes a lo que se está programando, únicamente se decida entre dos posibilidades que son mutuamente excluyentes, en esos casos basta con una estructura `if-else`. Por ejemplo, el siguiente código determina si un número es par o impar (un número entero cualquiera o es par o es impar, no hay otra posibilidad).

````python
n = 1000

if n%2 == 0:
    print("El número es par")
else:
    print("El número es impar")
````

La comprobación la hace al comprobar que el numero sea divisible entre 2 (es decir que el resto al dividir entre 2 sea 0).



Supongamos ahora otro caso en donde las posibilidades son mutuamente excluyentes: en un curso la calificación mínima para aprobar es de 70 y las calificaciones van de 0 a 100, ¿cómo podríamos implementar un código que decida y muestre si un alumno aprobó o suspendió la asignatura? Observa el siguiente código:

````python
calificacion = 100

if calificacion >= 70:
    print("Aprobado")
else:
    print("No aprobado")
````

Ahora, consideremos que en lugar de una clasificación binaria “Aprobado” y “No aprobado” quisiéramos utilizar una escala cualitativa como la siguiente:

- De 95 a 100 → Excelente
- De 85 a 94 →  Muy bien
- De 75 a 84 → Bien
- De 70 a 74 → Regular
- Menos de 70 →  Insuficiente

Para una situación como esta podemos utilizar la estructura `if-elif-else`:

````python
calificacion = 71

if calificacion >= 95 and calificacion <= 100:
    print("Excelente")
elif calificacion >= 85 and calificacion <= 94:
    print("Muy bien")
elif calificacion >= 75 and calificacion <= 84:
    print("Bien")
elif calificacion >= 70 and calificacion <= 74:
    print("Regular")
else:
    print("Insuficiente")
````

## Bucle `for`
El bucle `for` es una estructura de control de **repetición** que permite ejecutar un bloque de instrucciones varias veces, de manera automática, sobre una secuencia de elementos. Esta estructura se usa para recorrer listas, cadenas de texto, rangos de números o cualquier otro tipo de colección.

La sintaxis general de un bucle `for` en Python es:
```
for variable in secuencia:
  # hacer algo con cada elemento
```
Donde `variable` es el nombre que tomará, uno a uno, los valores de los elementos de la secuencia y `secuencia` puede ser, por ejemplo, una lista, una cadena o un rango de números.

Al igual que con los condicionales, la **indentación** es obligatoria en Python para indicar qué instrucciones están dentro del bucle.

### La función `range()`
La función `range()` nos permite generar secuencias de números enteros. Dependiendo de lo que queramos conseguir, podemos usar esta función de varias maneras:
- `range(valor_final)`: genera los números desde `0` hasta `valor_final - 1`.




````python
for numero in range(5):
  print(numero)
````

- `range(valor_inicial, valor_final)`: genera los números desde `valor_inicial` hasta `valor_final`.

````python
for numero in range(3, 10):
  print(numero)
````

- `range(valor_inicial, valor_final, paso)`: genera los números desde `valor_inicial` hasta `valor_final - 1`, cambiando de valor en cada paso según el valor del `paso`.

````python
for numero in range(0, 10, 2):
  print(numero)
````

El valor de `paso` puede ser positivo, si queremos que la secuecia de número vaya de menor a mayor, o negativo, si queremos que vaya de mayor a menor. Además, para ir de mayor a menor, `valor_inicial` tendrá que ser mayor que `valor_final`.

````python
for numero in range(10, 0, -1):
  print(numero)
````

Dentro del bucle `for` podemos hacer cualquier tipo de operaciones, no solo mostrar resultados por pantalla. En el siguiente ejemplo, el programa realiza la suma de todos los número desde el 1 hasta el 100:

````python
suma = 0

# Es habitual usar i como nombre para la variable del bucle
for i in range(101):
  suma += i

print(suma)
````

También podemos anidar estructuras de control, por ejemplo, un condicional dentro de un bucle. En el siguiente ejemplo, usaremos la estructura condicional para mostrar únicamente los número pares:

````python
for i in range(10):
  if i % 2 == 0:
    print(i)
````

### Bucles y cadenas

Las cadenas de texto (`str`) también son secuencias, por lo que se pueden recorrer carácter por carácter utilizando un bucle `for`.

````python
mensaje = "hola"

for letra in mensaje:
  print(letra)
````

De esta forma podemos acceder a cada carácter de la cadena y realizar las operaciones que necesitemos. En el siguiente ejemplo, se muestra cada letra de la palabra en mayúscula usando la función `.upper()`:

````python
nombre = "ana"

for letra in nombre:
  print(letra.upper())
````

##El bucle `while`
El bucle `while` es una estructura de control repetitiva que permite ejecutar un bloque de código mientras se cumpla una determinada condición. A diferencia del bucle `for`, que recorre una secuencia de elementos, el `while` se repite mientras la condición que se evalúa sea verdadera.

```
while condición:
  # hacer algo
```
`condición` debe ser una expresión que develv `True` o `False`, es decir, una odición lógica. Cuando la condición ea falsa, el bucle se detiene.

Como en los condicionales y otros bloques de código en Python, la **indentación** es obligatoria: todo lo que esté indentado debajo del `while` será el cuerpo del bucle.

````python
contador = 1

while contador <= 5:
  print("Contador:", contador)
  contador += 1
````

**¡Importante!** Hay que asgurarse de que la condición del bucle puede volverse falsa en algun momento, si no, el bucle nunca terminará. Esto es lo que se conoce como **bucle infinito**.

````python
# No lo ejecutes, es solo un ejemplo de bucle infinito
x = 1

while x > 0:
  print(x)
````

La condición del bucle puede ser simple, como una comparación, o se puede complicar combinando varias condiciones.
En el siguiente ejemplo, un juego para adivinar un número secreto, el bucle se repetirá hasta que el usuario acierte el número o hasta que supere el número de intentos permitido:

````python
numero_secreto = 7
intentos = 0
numero_usuario = 0

while numero_usuario != numero_secreto and intentos < 5:
    numero_usuario = int(input("Adivina el número (entre 1 y 10): "))
    intentos = intentos + 1

if numero_usuario == numero_secreto:
    print("¡Has acertado! 🎉")
else:
    print("¡Has agotado tus intentos! El número era", numero_secreto)
````

### La instrucción `break`
En Python, a veces necesitamos modificar el comportamiento normal de un bucle. La instrucción `break` sirve para salir del bucle inmediatamente, aunque la condición del `while` o el `for` siga siendo verdadera.


````python
intentos = 0

while True:
    respuesta = input("Escribe 'salir' para terminar: ")
    intentos = intentos + 1

    if respuesta == "salir":
        break  # sale del bucle

print("Has salido del bucle después de", intentos, "intento(s).")
````

### La instrucción `continue`
La instrucción `continue` sirve para saltar directamente al siguiente ciclo del bucle, ignorando lo que queda por ejecutar dentro del mismo ciclo actual.


````python
numero = 0

while numero < 5:
    numero = numero + 1

    if numero == 3:
        continue  # salta directamente a la siguiente vuelta

    print("Número:", numero)
````

## Control de excepciones

El control de excepciones en Python es un mecanismo que permite gestionar errores que pueden ocurrir durante la ejecución de un programa. Utiliza bloques `try, except, else y finally` para detectar y manejar estas situaciones de forma controlada. El bloque `try` contiene el código que puede generar una excepción, el bloque `except` define cómo manejarla, `else` se ejecuta si no ocurre ninguna excepción y `finally` se ejecuta siempre, ocurra o no una excepción. Esto evita que el programa se detenga inesperadamente y permite ofrecer respuestas más seguras y predecibles.

A continuación vamos a poner un par de ejemplos como se controlan las excepciones en un programa.

````python
try:
    # Intentamos dividir dos números
    numerador = int(input("Introduce el numerador: "))
    denominador = int(input("Introduce el denominador: "))
    resultado = numerador / denominador
except ZeroDivisionError:
    print("Error: No se puede dividir entre cero.")
except ValueError:
    print("Error: Debes introducir un número válido.")
else:
    print("El resultado de la división es: ", resultado)
````

Otro ejemplo de control de errores sería utilizando condicionales para comprobar que los datos introducidos no sean inválidos.
