# TUTORIAL DE USO DE UN NOTEBOOK
___

Un notebook (cuaderno interactivo) es una herramienta que permite escribir y ejecutar código junto a texto explicativo, todo en un mismo lugar.

Es muy utilizado en programación, ciencia de datos, educación y análisis de información, porque permite:

- Escribir código paso a paso y ver los resultados al instante.
- Añadir explicaciones en lenguaje natural (como este texto) usando formato Markdown.
- Combinar gráficos, fórmulas, tablas y código, ideal para informes o aprendizaje interactivo.

¿Y para qué se usa?

Con un notebook aprenderás a programar de forma práctica, analizar datos, hacer gráficos y documentar experimentos de código. Es como una libreta digital que ejecuta lo que escribes y te explica lo que haces, todo al mismo tiempo.

¿Y cómo se usa?

Un notebook permite ver el código fuente simplemente clicando dos veces sobre el texto que ves. Haz la prueba si quieres.








# 1. Bloques
___

Un notebook tiene normalmente dos bloques bien diferenciados que se pueden intercalar: bloques de texto y de código. Para añadirlos simplemente deberemos hacer lo siguiente. Cuando tengamos abierto Pycharm veremos el siguiente grupo de botones:

- **Code** : nos permitirá añadir bloques de código. Lo utilizaremos para poner los fragmentos de los programas que vayamos realizando.
- **Markdown** : nos permitirá añadir bloques de texto y en ellos incluiremos explicaciones, ejemplos, etc.


# 2. Formato
___


A lo largo de un notebook podemos poner diferente tamaño y estilo a lo que escribimos (desde encabezados, negrita, cursiva, hasta imágenes).

Con el símbolo # podemos poner encabezados. A más # seguidos pongamos, el encabezado será un nivel superior (es decir, más pequeño).

# Esto es un encabezado de nivel 1

## Esto es un encabezado de nivel 2

Podemos poner énfasis en ciertas palabras poniendolas en **negrita** o en *cursiva*.

Lo parrafos se separan con líneas vacías.

* Podemos crear listas.
* Las cuales se crean usando asteriscos.

1. También se pueden hacer numeradas.
2. Si queremos que estén ordenadas.

[Incluso podemos poner hipervículos](https://www.youtube.com/watch?v=xvFZjo5PgG0)

Podemos en un bloque de texto meter palabras con formato de código usando una sola tilde invertida: `foo()`, y bloques de código usando 3 tildes invertidas:
```
bar()
```
También le podemos poner sangría al código si incluimos 4 espacios:
```
    foo()
```

# 3. Kernels
___

Detrás de cada notebook se ejecuta un núcleo (kernel). Cuando ejecutas un bloque de código, ese código se ejecuta dentro del kernel y se devuelve un resultado para mostrarse.
El estado del kernel se mantiene a lo largo del tiempo y entre bloques, de tal forma que está relacionado con todo el documento, no solo con bloques individuales.

A continuación vamos a ver un ejemplo:




````python
num1 = 2
num2 = 4
````

Como podemos ver, hemos ejecutado este bloque sin ningún tipo de problema. Si ahora ejecutamos el siguiente bloque, el resultado esperado será 5.

````python
num1+num2
````

Pero ¿qué pasaría si hubiesemos ejecutado primero el segundo bloque de código en vez del primero? Simplemente hubiese salido un error.

También hay que decir, que si modificamos algo de una parte del código, para que los cambios sean visibles hay que volver a ejecutarlo todo desde la parte modificada, sino los cambios no serán visibles y se seguirán usando los datos previamente introducidos.
